### Asset management

total_asset * risk = position_size * leverage * stoploss

```python
def position_size(stoploss, risk=0.015, leverage=10, total_asset=100):
    return f"{((total_asset * risk) / (leverage * stoploss)):.2f}$"
```

### Risk management



**Maximum risk** = 1.5%
1. Very high risk: have a sub-account (for excitements or few confirmations for example 2 out of 10) -> 0.1% to 0.25%
2. High risk: in main account (all confirmed but trigger is not executed) -> 0.25% to 0.5%
3. Medium risk: in main account (most of confirmations for example 6-7 out of 10 and also trigger) -> 0.5% to 0.75%
4. Low risk: -> in main account 0.75% to 1%

**Note:** The 0.5% remained for backup

**Winrates: (35%)**
1. High risk: about 20%
2. Medium risk: about 40%
3. Low risk: about 65%
