```python
import requests
import json
```

```python
with open("../secrets.json", mode='r') as file:
    secrets = json.load(file)

token = secrets.get("Abantether")
```

```python
def get_balance(token):
    url = "https://abantether.com/api/v1/users/balance/"
    headers = {
        "Authorization": "Token " + token,
        "content-type": "application/json"
    }
    res = requests.get(url, headers=headers)
    return res.json()
```

```python
get_balance(token)
```

```python
def get_ticker_data(ticker=None):
    url = "https://api.abantether.com/api/v1/manager/otc/ticker"
    if ticker: url += f"?coin={ticker}"
    res = requests.get(url)
    return res.json()
```

```python
get_ticker_data("USDT")
```

```python
def get_whitelist(token):
    url = "https://api.abantether.com/transaction_bc/whitelist"
    headers = {
        "Authorization": token
    }
    res = requests.get(url, headers=headers)
    return res.json()
```

```python
get_whitelist(token)
```
