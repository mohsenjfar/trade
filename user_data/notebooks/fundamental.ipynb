```python
import os
from coingecko_sdk import Coingecko
from dotenv import load_dotenv
import pandas as pd

load_dotenv()
```

```python
client = Coingecko(
    demo_api_key=os.environ.get("COINGECKO_DEMO_API_KEY"),
    environment="demo"
)
```

```python
price_data = client.simple.price.get(
    vs_currencies="usd",
    ids="bitcoin",
)
```

```python
price_data['bitcoin'].to_dict()
```

```python
cats = client.coins.categories.get_list()
platforms_response = client.asset_platforms.get()
markets_response = client.coins.markets.get(vs_currency='usd')
```

```python
dataframe = pd.DataFrame([cat.to_dict() for cat in cats])
dataframe
```

```python
platforms_dataframe = pd.DataFrame(plat.to_dict() for plat in platforms_response)
platforms_dataframe
```

```python
platforms_dataframe = pd.DataFrame(market.to_dict() for market in markets_response)
platforms_dataframe
```
